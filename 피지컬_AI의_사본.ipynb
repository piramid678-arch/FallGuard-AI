```python
!pip install gymnasium[mujoco]
import gymnasium

# 환경 생성
env = gymnasium.make("Humanoid-v5", render_mode="rgb_array")

# 환경 초기화
observation, info = env.reset()

print("시뮬레이션 시작...")

for _ in range(100):
    # 무작위 액션 선택
    action = env.action_space.sample()

    # 환경 실행 (1 스텝)
    observation, reward, terminated, truncated, info = env.step(action)

    # 에피소드가 끝나면 리셋
    if terminated or truncated:
        observation, info = env.reset()
        print("에피소드 종료 및 리셋")

print("시뮬레이션 완료")
env.close()
```
